# 차량사고 감지에 사용된 접근 방식의 개요

우리는 정적인 이미지에 차량 사고가 나타나 있는지를 식별할 수 있는 애플리케이션을 개발하는 과제를 맡았습니다.  
만약 사고가 감지되면, 해당 청구는 적절한 청구 처리 부서로 전달됩니다.  
우선 YOLO 모델을 어떻게 사용하는지부터 알아보겠습니다.

Note: YOLO는 <I>'You Only Look Once'</I>의 약자로, 해당 모델에서 사용되는 알고리즘의 별칭입니다.

이 노트북에서는 사전 학습된 머신러닝 모델인 YOLOv8m을 사용하며, 이 모델이 정적인 이미지에서 어떻게 작동하는지 살펴봅니다.  
만약 자동차를 감지할 수 있다면, 해당 모델을 차량 사고를 감지하도록 재학습시키는 것도 시도해볼 수 있겠죠?

먼저, YOLO 모델을 다루는 데 사용되는 `ultralytics` 패키지와 기타 필요한 Python 패키지들을 임포트하겠습니다.  
`ultralytics` 패키지를 설치하면 YOLOv8 모델을 간단한 명령어로 불러와서 쉽게 실행할 수 있습니다. 

```python
# 이 실습을 위해 사전구성된 워크벤치 이미지를 사용하지 않았다면, 아래 줄의 주석을 해제하고 실행하여 필요한 패키지들을 설치할 수 있습니다.
# !pip install --no-cache-dir --no-dependencies -r requirements.txt

# YOLO 및 이미지 처리 패키지 불러오기
import cv2
from ultralytics import YOLO  # YOLOv8 모델을 사용할 수 있는 라이브러리
from PIL import Image  # 이미지 로딩 및 표시를 위한 라이브러리
```

```python
# 이번 실습에서는 객체 감지를 위해 YOLOv8m 모델을 사용할 것입니다.
model = YOLO('yolov8m.pt')  # 모델 불러오기
```

```python
# 테스트할 이미지를 화면에 표시합니다.
Image.open("images/carImage0.jpg")
```

```python
# YOLO 모델이 주어진 사진에서 자동차 객체를 식별할 수 있는지 확인합니다.
results = model.predict("images/carImage0.jpg")  # 이미지에 대한 예측 수행
```

YOLO는 제공된 사진에서 '자동차 1대'를 감지한 것으로 나타났습니다.  
이는 매우 긍정적인 결과입니다. 이제 YOLO가 자동차 객체를 제대로 감지하는지 확인하기 위해 몇 장의 사진을 더 테스트해보겠습니다.

```python
# <images> 폴더에 있는 모든 이미지를 나열합니다.  
# 테스트에 사용할 이미지를 그 중에서 선택할 수 있습니다.  
!ls -al images
```

```python
# 예를 들어 <carImage4.jpg>와 같은 자동차 이미지를 선택하여, YOLO가 해당 이미지에서 자동차를 식별할 수 있는지 확인합니다.  
img = Image.open('images/carImage4.jpg')
img_resized = img.resize((500, 300))
display(img_resized)
```

```python
# 모델 예측 결과를 가져옵니다.
results = model.predict(img)
```

<carImage4.jpg>의 결과를 보면, YOLO는 총 17대의 자동차와 1대의 트럭이 있다고 판단했습니다.  
이 결과는 흥미롭습니다. YOLO 모델이 여러 대의 차량이 포함된 이미지에서도 여러 대의 자동차를 식별할 수 있음을 확인했습니다.

하지만 이미지에 너무 많은 차량이 표시되어 있을 경우, 사람인 우리는 정확히 몇 대의 차량이 있는지 세기 어려울 수 있습니다.

따라서 YOLO 모델이 이미지 내에서 감지한 자동차 주위에 상자를 그려주는 코드를 작성해보겠습니다.
