## Jupyter 노트북이란?

이 문서는 [Jupyter](https://jupyter.org/) 노트북입니다. 시작하기 위해 알아야 할 사항은 몇 가지뿐입니다.

* Jupyter 노트북은 _셀(cell)_ 로 구성되어 있으며, 셀에는 설명 텍스트, 실행 가능한 코드, 또는 인터랙티브한 UI 요소가 포함될 수 있습니다. 현재 보고 계신 셀은 설명 텍스트 셀입니다.

* 셀은 수정하거나 실행할 수 있습니다. 셀을 수정하려면 셀을 더블 클릭하여 테두리가 활성화되도록 합니다. 셀을 실행하려면, 화면 상단의 ▶️ 버튼을 클릭하거나 셀을 선택한 후 SHIFT+ENTER 키를 누르면 됩니다.

     <figure>
        <img src="../../content/modules/ROOT/assets/images/02/02-04-play-button.png"  alt='missing' width="300"  >
     <figure/>

    Jupyter는 셀 안의 코드를 실행하고, 실행 결과를 기록한 뒤 다음 셀로 이동합니다.

* 셀이 실행 중일 때는 셀 왼쪽에 별표(`*`)가 표시됩니다. 예: `In: [*]`  
  실행이 완료되면 별표는 순서 번호로 바뀝니다.

이제 직접 실습해 보겠습니다. 다음 셀을 실행해 주십시오.

```python
def print_some_text(entered_text):
    print("입력한 내용은 다음과 같습니다: \n" + entered_text)


my_text = "이 셀은 코드를 위한 공간입니다. 위의 재생 버튼을 클릭하거나 SHIFT+ENTER를 눌러 코드를 실행할 수 있습니다."

print_some_text(my_text)
```

```python
new_text = "다음으로, 서술 텍스트나 문서를 어떻게 작성할지 살펴보겠습니다."

# previously defined functions from another cell are still accessible.
print_some_text(new_text)
```
